# Grafica una serie de Fourier y las componentes

```python
%matplotlib inline
from matplotlib import pyplot as plt
import numpy as np
from ipywidgets import interact
```

```python
def Serie(n):
    suma = 0
    for k in range(1, n, 2):
        def s(t):
            return 4*np.sin(100*np.pi*k*t)/(k*np.pi)
        t = np.linspace(0, 2/100., 1000)
        suma += s(t)
    plt.plot(t, suma, 'k')
```

```python
interact(Serie, n = (2, 20, 2))
```

```python
def Sintesis(n):
    suma = 0
    for k in range(1, n, 2):
        def s(t):
            return 4*np.sin(100*np.pi*k*t)/(k*np.pi)
        t = np.linspace(0, 2/100., 1000)
        suma += s(t)
        plt.plot(t, s(t), 'k')
```

```python
interact(Sintesis, n = (2, 20, 2))
```
