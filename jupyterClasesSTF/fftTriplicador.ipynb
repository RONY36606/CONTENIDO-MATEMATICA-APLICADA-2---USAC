# Tripicador de frecuencia de una senoide

### Códigos para triplicar la frecuencia $f$ de la señal $x(t) = \sin(2\pi ft)$.

##### Si $f = \frac{1}{4} Hz, x(t) = \sin(\frac{\pi}{2}t)$

```python
import numpy as np
import matplotlib.pyplot as plt
dt = 0.1 #seg
t = np.arange(0.0, 100.0, dt)
T1 = 4.0 # seg, períod0
f1 = 1/T1 # frecuencia
s = np.sin(2*np.pi*f1*t)
plt.figure(figsize = (20, 3))
plt.plot(t, s)
```

##### Construcción de una señal cuadrada con la misma frecuencia de $x(t)$ original.

```python
#import numpy as np
#import matplotlib.pyplot as plt
#dt = 0.1 #seg
#t = np.arange(0.0, 100.0, dt)
#T1 = 4.0 # seg
#f1 = 1/T1
x = np.floor(1 + 0.8*s) # x(t) = 0.8*np.sin(2*np.pi*f1*t))
plt.figure(figsize = (20, 3))
plt.plot(t, x)
```

##### Espectro en la frecuencia de la onda cuadrada $f1 = 0.25, f3 = 0.75, f5 = 1.25$ Hz

```python
X = np.fft.fft(x)
freq = np.fft.fftfreq(np.size(X), d = dt)
plt.figure(figsize = (20, 3))
plt.plot(freq, np.abs(X), "r")
```

##### La fundamental y tercer armónica, $f_1 = 0.25, f_3 = 0.75 $ Hz

```python
X = np.fft.fft(x)
freq = np.fft.fftfreq(np.size(X), d = dt)
plt.figure(figsize = (20, 3))
plt.plot(freq, np.abs(X), "r")
plt.vlines(f1, -100, 600, "k")
plt.vlines(3*f1, -100, 600, "b")
```

##### Filtro de la tercera armónica $f3 = 0.75$ Hz

```python
X1 = np.copy(X)
X1[np.abs(freq) < 0.75] = 0
X1[np.abs(freq) > 0.75] = 0
plt.figure(figsize=(20,3))
plt.plot(freq, np.abs(X1), "r")
```

##### La señal senoidal con frecuencia triplicada respecto a la señal original.

```python
x1 = np.real(np.fft.ifft(X1))*5
plt.figure(figsize=(20,3))
plt.plot(t, x1, "k")
```

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
dx = 0.1 # seg
x = np.arange(0.0, 100.0, dx)
l1 = 4.0 # segs
v1 = 1/l1 # Hzs
l2 = 1.0 # segs
v2 = 1/l2 # Hzs
f = np.sin(2*np.pi*v1*x) + np.sin(2*np.pi*v2*x)
plt.figure(figsize=(20, 3))
plt.plot(x, f)
```

```python
F = np.fft.fft(f)
freq = np.fft.fftfreq(np.size(F), d = dx)
plt.figure(figsize=(20,3))
plt.plot(freq, np.abs(F), "r")
```

```python
F = np.fft.fft(f)
freq = np.fft.fftfreq(np.size(F), d = dx)
plt.figure(figsize=(20,3))
plt.plot(freq, np.abs(F), "r")
plt.vlines(v1, -100, 600, "k")
plt.vlines(v2, -100, 600, "k")
```

```python
f1 = np.real(np.fft.ifft(F))
plt.figure(figsize=(20,3))
plt.plot(x, f1)
```

```python
F1 = np.copy(F)
F1[np.abs(freq)>0.5]=0
plt.figure(figsize=(20,3))
plt.plot(freq, np.abs(F1), "r")
```

```python
f1 = np.real(np.fft.ifft(F1))
plt.figure(figsize=(20,3))
plt.plot(x, f1, "k")
```
