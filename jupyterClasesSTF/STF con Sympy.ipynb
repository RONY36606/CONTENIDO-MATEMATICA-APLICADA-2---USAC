### Calcula la serie para una sola fórmula por ejemplo la diente de sierra impar: $f(x) = x, -\pi<x< \pi\;\;\; \; T= 2\pi$ 

```python
from sympy import fourier_series, pi
from sympy.abc import x
s = fourier_series(x, (x, -pi, pi))  # aqui desde -T/2 a T/2
s.scale(2).truncate(5)
```

#### Gráfica de series truncadas

```python
from sympy import fourier_series, pi, plot
from sympy.abc import x
f = x
s = fourier_series(f, (x, -pi, pi))
s1 = s.truncate(n = 3)
s2 = s.truncate(n = 5)
s3 = s.truncate(n = 7)
p = plot(f, s1, s2, s3, (x, -pi, pi), show=False, legend=True)
p[0].line_color = (0, 0, 0)
p[0].label = 'x'
p[1].line_color = (0.7, 0.7, 0.7)
p[1].label = 'n=3'
p[2].line_color = (0.5, 0.5, 0.5)
p[2].label = 'n=5'
p[3].line_color = (0.3, 0.3, 0.3)
p[3].label = 'n=7'
p.show()
```

### Serie de Fourier de onda triangular par, por partes: $f(t)= \displaystyle \left\{  { -t, \hspace{2mm} -\pi \leq t\leq 0 \atop t, \hspace{9mm} 0 <  t < \pi } \right.$     $\;\; T = 2\pi$

```python
import sympy as sym
from sympy import fourier_series, pi, plot
from sympy.abc import t
ft = sym.Piecewise((-t, t < 0),
                   (t, t >  0))
s = fourier_series(ft, (t, -pi, pi))  #aqui desde -T/2 a T/2
s.scale(1).truncate(5)
```

### Gráfica series truncadas

```python
s = fourier_series(ft, (t, -pi, pi))
s1 = s.truncate(n = 3)
s2 = s.truncate(n = 5)
s3 = s.truncate(n = 7)
p = plot(pi/2 + 0*t, s1, s2, s3, (t, -pi, pi), show=False, legend=True)
p[0].line_color = (0, 0, 0)
p[0].label = 't'
p[1].line_color = (0.7, 0.7, 0.7)
p[1].label = 'n=3'
p[2].line_color = (0.5, 0.5, 0.5)
p[2].label = 'n=5'
p[3].line_color = (0.3, 0.3, 0.3)
p[3].label = 'n=7'
p.show()
```

### Calculo de la STF de $f(t)= \displaystyle \left\{  { -3t, \hspace{2mm} -1\leq t\leq 0 \atop t, \hspace{9mm} 0 <  t < 1 } \right.$     $\;\; T = 2$

```python
import sympy as sym
from sympy import fourier_series, pi
from sympy.abc import t
ft = sym.Piecewise((-3*t, t < 0),
                   (t, t >  0))
s = fourier_series(ft, (t, -1, 1))  #aqui desde -T/2 a T/2
s.scale(1).truncate(7)
```

### Calculo de la STF de$ f(x)= \left\{ \begin{array}{lcc}
             0 &   si  & -4 < x \leq -2 \\
             \\ 1 &  si & -2 < x < 2 \\
             \\ 0 &  si  & 2 < x \leq 4
             \end{array}
   \right.$  $\hspace{1cm}$  Con $T = 8$

```python
import sympy as sym
from sympy import fourier_series, pi
from sympy.abc import t
ft = sym.Piecewise((0, t < -2),
                   (1, t <  2),
                   (0, True))
s = fourier_series(ft, (t, -4, 4))  # aqui desde -T/2 a T/2
s.scale(1).truncate(5)
```
