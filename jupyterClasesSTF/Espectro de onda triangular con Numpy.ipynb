### Generación de Onda Triangular y calculo de su espectro en la frecuencia con Numpy

##### Es necesario saber como generar la onda triangular usando la función piso (sugerencia grafique la función floor), librerías y comandos necesaros en Numpy.

```python
import numpy as np
import matplotlib.pyplot as plt
dt = 0.1 #seg
t = np.arange(0.0, 10.0, dt)
```

#### Gráfica en el tiempo $0\leq t\leq 10$ seg

```python
y = 0.5*(2*(t - np.floor(t)) - 1)*(-1)**np.floor(t) + 0.5
plt.figure(figsize = (20, 3))
plt.plot(t, y)
```

#### Gráfica en el tiempo $0 \leq t \leq 100$ seg

```python
import numpy as np
import matplotlib.pyplot as plt
dt = 0.1 #seg
t = np.arange(0.0, 100.0, dt)
T1 = 2.0 # seg, períod0
f1 = 1/T1 # frecuencia fundamental
tri = 0.5*(2*(t - np.floor(t)) - 1)*(-1)**np.floor(t) + 0.5
plt.figure(figsize = (20, 3))
plt.plot(t, tri)
```

#### Grafica del espectro de amplitud usando las librerías fft y fftfreq de Numpy, $f_0 = \frac{1}{2}$ cps

```python
Tri = np.fft.fft(tri)
freq = np.fft.fftfreq(np.size(Tri), d = dt)
plt.figure(figsize = (20, 3))
plt.plot(freq, np.abs(Tri), "r")
```
