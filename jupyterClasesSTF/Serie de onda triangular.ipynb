# Cálculo de Serie trig Onda Triangular 

```python
# librerías
from sympy import*
```

```python
t = symbols("t")
n = symbols("n", int=True, positive=True)
T0 = 2*pi                                         #  ingrese período período
m = 7                                             # ingreses número de componente
# construya la función por partes
ft = Piecewise((-t, t <=0), (t, t <= pi), (0, True))
#frecuencia angular
w0 = 2*pi/T0  
```

```python
a0 = (1/T0*integrate(ft, (t, -T0/2, T0/2)))   #note a0 representa a0/2
a0
```

```python
an = (2/T0)*integrate(ft*cos(w0*n*t), (t, -T0/2, T0/2))
simplify(an)
```

```python
bn = (2/T0)*integrate(ft*sin(w0*n*t), (t, -T0/2, T0/2))
simplify(bn)
```

```python
ai = [an.subs(n, k) for k in range(1, m)]
bi = [bn.subs(n, k) for k in range(1, m)]
print('n       an       bn')
print('%d %8s       --'%(0, N(a0, 3)))
for i in range(m-1):
    print ('%d %8s %8s' %(i+1, N(ai[i], 3), N(bi[i], 3)))
```

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib as mpl

fig, g = plt.subplots() 
T0n = 2*np.pi #*np.pi             #hay que redefinir el periodo para numpy
w0n = 2*np.pi/T0n
tm = np.linspace(0, 2*T0n, 100)

for k in range(m-1):
    g.plot(tm, ai[k]*np.cos(w0n*(k +1)*tm) + bi[k]*np.sin(w0n*(k + 1)*tm))  #plot de armónica
g.plot(tm, a0 + 0*tm)         #plot de a0/2
plt.xlabel('t, seg')
plt.ylabel('amplitudes')
plt.title('Armonicas')
plt.grid(True)
plt.show()
#para plotear la serie de Fourier
s = [0]*100
for k in range(m-1):
    tm = np.linspace(0, 2*T0n, 100)
    sp = (ai[k]*np.cos(w0n*(k + 1)*tm) + bi[k]*np.sin(w0n*(k + 1)*tm))
    s += sp
s = s + 3/4
plt.plot(tm, s, 'k')
plt.title('x(t)')
plt.xlabel('t, seg')
plt.ylabel('x(t)')
plt.grid(True)
plt.show()
```

```python
ai0 = [a0]; bi0 = [0]  #para graficar el espectro en n
ai0.extend(ai); bi0.extend(bi)
for i in range(m):
    x = [[i, i]]
    y = [[0, 0.5*abs((ai0[i])**2 + (bi0[i])**2)]]
    plt.plot(x[0], y[0])
plt.title('Espectro de amplitud')
plt.xlabel('f(Hz)')
plt.ylabel('|A|')
plt.grid(True)
#plt.xlim(0,5)
#plt.ylim(0,5)
plt.show
```
