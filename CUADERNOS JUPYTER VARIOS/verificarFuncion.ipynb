```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def analizar_simetria_senal(t, x, T=None, tol=1e-3):
    """
    Evalúa las simetrías de una señal dada.
    
    Parámetros:
    - t: Vector de tiempo (debe estar centrado en 0 para evaluar paridad).
    - x: Vector de valores de la señal.
    - T: Período fundamental de la señal.
    - tol: Tolerancia numérica para la comparación de arreglos.
    """
    resultados = {}
    
    # 1. Comprobación de tiempo centrado en cero
    es_tiempo_centrado = np.allclose(t, -np.flip(t), atol=tol)
    
    if es_tiempo_centrado:
        x_invertida = np.flip(x)
        es_par = np.allclose(x, x_invertida, atol=tol)
        es_impar = np.allclose(x, -x_invertida, atol=tol)
    else:
        es_par = False
        es_impar = False
        print("Aviso: El vector de tiempo no está centrado en t=0; no se evaluó la paridad estricta.")

    # 2. Comprobación de simetría de media onda
    es_media_onda = False
    if T is not None:
        dt = t[1] - t[0]
        desplazamiento_muestras = int(np.round((T / 2) / dt))
        
        if desplazamiento_muestras < len(x):
            x_original = x[:-desplazamiento_muestras]
            x_desplazada = x[desplazamiento_muestras:]
            es_media_onda = np.allclose(x_original, -x_desplazada, atol=tol)

    # 3. Combinaciones
    es_media_onda_par = es_media_onda and es_par
    es_media_onda_impar = es_media_onda and es_impar

    resultados = {
        'Par': es_par,
        'Impar': es_impar,
        'Simetría de Media Onda': es_media_onda,
        'Media Onda Par': es_media_onda_par,
        'Media Onda Impar': es_media_onda_impar
    }
    
    return resultados
```

```python
# Configuración del eje de tiempo centrado en 0
f0 = 1/20        # Frecuencia de 1 Hz
T = 1.0 / f0    # Período fundamental
fs = 1000       # Frecuencia de muestreo
dt = 1 / fs
t = np.arange(-2*T, 2*T, dt)

# Definición de señales de prueba
senales = {
    "Coseno (Par y Media Onda Par)": np.cos(2 * np.pi * f0 * t),
    "Seno (Impar y Media Onda Impar)": np.sin(2 * np.pi * f0 * t),
    "Coseno con Offset DC (Par, Sin Media Onda)": 1.5 + np.cos(2 * np.pi * f0 * t),
    "Onda Cuadrada Centrada (Media Onda Impar)": np.sign(np.sin(2 * np.pi * f0 * t))
}

# Evaluación de cada señal
for nombre, x in senales.items():
    print(f"=== {nombre} ===")
    res = analizar_simetria_senal(t, x, T=T)
    for propiedad, valor in res.items():
        print(f"  - {propiedad}: {valor}")
    print()
```
