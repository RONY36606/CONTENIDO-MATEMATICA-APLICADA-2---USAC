# Cálculo de Integrales Indefinidas y Definidas

```python
import sympy as sp
from IPython.display import display, Math

# 1. Variables simbólicas
t = sp.Symbol('t', real=True)
n = sp.Symbol('n', integer=True, positive=True)

# 2. Función
f = (8/20) * ((1 - t) * sp.cos((2 * sp.pi * n * t) / 20))

# 3. Integral definida de 0 a 1 (sin la 't' extra)
resultado_raw = sp.integrate(f, (t, 0, 1))

# 4. Simplificación
resultado_simplificado = sp.simplify(resultado_raw)

# Mostrar resultados
display(Math(r'\text{Función: } f(t) = ' + sp.latex(f)))
display(Math(r'\text{Resultado bruto: } ' + sp.latex(resultado_raw)))
display(Math(r'\text{Resultado simplificado: } ' + sp.latex(resultado_simplificado)))
```

```python
import sympy as sp
from IPython.display import display, Math

# 1. Variables simbólicas
t = sp.Symbol('t', real=True)
n = sp.Symbol('n', integer=True, positive=True)

# 2. Función
f = (1/2) * ((-3*t))

# 3. Integral definida de 0 a 1 (sin la 't' extra)
resultado_raw = sp.integrate(f, (t, -1, 0))

# 4. Simplificación
resultado_simplificado = sp.simplify(resultado_raw)

# Mostrar resultados
display(Math(r'\text{Función: } f(t) = ' + sp.latex(f)))
display(Math(r'\text{Resultado bruto: } ' + sp.latex(resultado_raw)))
display(Math(r'\text{Resultado simplificado: } ' + sp.latex(resultado_simplificado)))
```
