```python
import sympy as sp

def inversa_fourier_fracciones_parciales(F_w, nombre=""):
    w = sp.Symbol('w', real=True)
    t = sp.Symbol('t', real=True)
    s = sp.Symbol('s')
    
    # 1. Cambio de j*w por s
    F_s = F_w.subs(sp.I * w, s)
    
    # 2. Fracciones parciales
    F_frac_s = sp.apart(F_s, s)
    F_frac_w = F_frac_s.subs(s, sp.I * w)
    
    # 3. Inversa de Fourier
    f_t = sp.inverse_laplace_transform(F_frac_s, s, t, noconds=True)
    f_t_limpia = sp.collect(f_t, sp.Heaviside(t))
    
    # Impresión en 1 sola línea por paso (evita el truncado)
    print("=" * 65)
    print(f" {nombre} ")
    print("=" * 65)
    print(f"1. F(ω) original:     {F_w}")
    print(f"2. Fracciones en s:   {F_frac_s}")
    print(f"3. Fracciones en jω:  {F_frac_w}")
    print(f"4. Resultado f(t):    {f_t_limpia}")
    print("=" * 65 + "\n")

if __name__ == "__main__":
    w = sp.Symbol('w', real=True)
    
    # Ejercicio 2
    F2 = 1 / ((1 + sp.I*w) * (1 + 2*sp.I*w))
    inversa_fourier_fracciones_parciales(F2, "EJERCICIO 2 DE LA GUÍA")
    
    # Ejercicio 3
    F3 = (sp.I * w) / ((1 + sp.I*w) * (1 + 2*sp.I*w))
    inversa_fourier_fracciones_parciales(F3, "EJERCICIO 3 DE LA GUÍA")
```
