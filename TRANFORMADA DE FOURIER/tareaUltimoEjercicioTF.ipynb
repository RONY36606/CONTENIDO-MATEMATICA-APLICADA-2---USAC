```python
import numpy as np
import matplotlib.pyplot as plt

# Definir tiempo t y la señal original
t_vals = np.linspace(-4, 4, 1000)
f_original = np.maximum(0, 1 - np.abs(t_vals))  # Triángulo original

# 4b & 4c. Integración numérica de la señal filtrada en [-2pi, 2pi]
w_vals = np.linspace(-2 * np.pi, 2 * np.pi, 1000)
dw = w_vals[1] - w_vals[0]

f_filtrada = []
for t in t_vals:
    # F(w) para el pulso triangular (evaluación segura en w=0)
    F_w = np.where(w_vals == 0, 1.0, 2 * (1 - np.cos(w_vals)) / (w_vals**2))
    # Transformada Inversa de Fourier en el rango limitado [-2pi, 2pi]
    integrando = F_w * np.cos(w_vals * t)
    val_t = (1 / (2 * np.pi)) * np.sum(integrando) * dw
    f_filtrada.append(val_t)

# Graficar la comparación
plt.figure(figsize=(9, 5))
plt.plot(t_vals, f_original, 'b--', lw=2, label=r'Señal Original $f(t)$')
plt.plot(t_vals, f_filtrada, 'r-', lw=2, label=r'Señal Filtrada $f_{filtrada}(t)$ ($|\omega| \leq 2\pi$)')
plt.title('Inciso 4c: Comparación de la Señal Original vs. Filtrada', fontweight='bold')
plt.xlabel('Tiempo (t)')
plt.ylabel('Amplitud')
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()
plt.show()
```
